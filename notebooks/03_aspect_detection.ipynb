# 03_aspect_detection.ipynb

Train a multi‑label aspect detection model using TF‑IDF features and a One‑vs‑Rest classifier.


```python
import warnings
warnings.filterwarnings('ignore')
```

```python
import sys
from pathlib import Path

# Ensure project root is in sys.path so 'src' is always importable
ROOT_DIR = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
if str(ROOT_DIR) not in sys.path:
    sys.path.insert(0, str(ROOT_DIR))

import pandas as pd
import joblib
from src.train import build_feature_pipeline, generate_weak_labels, train_aspect_detector

# Load preprocessed reviews
csv_path = ROOT_DIR / 'data' / 'processed' / 'reviews_preprocessed.csv'
df = pd.read_csv(csv_path)
texts = df['review_processed'].fillna('').astype(str).tolist()

# Weak labels based on keyword lexicon
Y = generate_weak_labels(texts)

# Train the aspect detector (default SVM)
vectorizer, model = train_aspect_detector(texts, Y)

# Ensure models directory exists and save artifacts
models_dir = ROOT_DIR / 'models'
models_dir.mkdir(parents=True, exist_ok=True)

joblib.dump(vectorizer, models_dir / 'tfidf_vectorizer.joblib')
joblib.dump(model, models_dir / 'aspect_model.joblib')
print(f'Aspect detection model and vectorizer saved to {models_dir}')
```
