# 03_aspect_detection.ipynb

Train a multi‑label aspect detection model using TF‑IDF features and a One‑vs‑Rest classifier.


```python
import pandas as pd
import joblib
from src.train import build_feature_pipeline, generate_weak_labels, train_aspect_detector

# Load preprocessed reviews
df = pd.read_csv('../data/processed/reviews_preprocessed.csv')
texts = df['review_processed'].tolist()

# Weak labels based on keyword lexicon
Y = generate_weak_labels(texts)

# Train the aspect detector (default SVM)
vectorizer, model = train_aspect_detector(texts, Y)

# Save artifacts
joblib.dump(vectorizer, '../models/tfidf_vectorizer.joblib')
joblib.dump(model, '../models/aspect_model.joblib')
print('Aspect detection model and vectorizer saved to models/')
```
