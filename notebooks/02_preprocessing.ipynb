Preprocess the raw reviews and save a cleaned version.


```python
import warnings
warnings.filterwarnings('ignore')
```

```python
import sys
from pathlib import Path

# Ensure project root is in sys.path so 'src' is always importable
ROOT_DIR = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
if str(ROOT_DIR) not in sys.path:
    sys.path.insert(0, str(ROOT_DIR))

import pandas as pd
from src.preprocessing import preprocess_batch

# Load cleaned CSV from previous step
csv_path = ROOT_DIR / 'data' / 'processed' / 'reviews_cleaned.csv'
df = pd.read_csv(csv_path)

# Support either 'text' or 'review' column name
text_col = 'text' if 'text' in df.columns else 'review'
print(f'Using text column: {text_col} ({len(df):,} rows)')

# Apply preprocessing to the review text
processed = preprocess_batch(df[text_col].fillna('').astype(str).tolist())
df['review_processed'] = processed

# Save for the next stage
out_path = ROOT_DIR / 'data' / 'processed' / 'reviews_preprocessed.csv'
df.to_csv(out_path, index=False)
print(f'Saved preprocessed reviews to {out_path}')
```
